### Тапсырма 20. Кібісе жыл

`year` айнымалысындағы жылдың кібісе жыл екенін анықтаңыз. Жыл 4-ке бөлінсе және 100-ге бөлінбесе, немесе 400-ге бөлінсе, ол кібісе жыл болып саналады. `2024`, `1900` және `2000` жылдарымен тексеріңіз.

**Күтілетін нәтиже:**

```
2024 - кібісе жыл
1900 - кібісе жыл емес
2000 - кібісе жыл
```

````python
years = [2024, 1900, 2000]
````

````python
for year in years:
    # Шарт: Жыл 4-ке бөлініп, 100-ге бөлінбесе ЖӘНЕ 400-ге бөлінсе
    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        print(f"{year} - кібісе жыл")
    else:
        print(f"{year} - кібісе жыл емес")
````

### Тапсырма 22. Үшбұрыштың түрі

Үшбұрыштың үш қабырғасының ұзындығы `a`, `b`, `c` берілген. Алдымен мұндай үшбұрыштың болуы мүмкін екенін тексеріңіз: кез келген екі қабырғаның қосындысы үшінші қабырғадан үлкен болуы керек. Үшбұрыш бар болса, оның тең қабырғалы, теңбүйірлі немесе әртүрлі қабырғалы екенін анықтаңыз.

````python
a, b, c = 4, 5, 5
````

````python
if (a + b > c) and (a + c > b) and (b + c > a):
  if a == b and b == c:
        print("Үшбұрыш бар: тең қабырғалы")
  elif a == b or b == c or a == c:
        print("Үшбұрыш бар: теңбүйірлі")
  else:
        print("Үшбұрыш бар: әртүрлі қабырғалы")
else:
    print("Мұндай үшбұрыш болмайды")
````

### Тапсырма 23. Қарапайым калькулятор

````python
a = 10
b = 0
operation = "/"
````

````python
if operation == "+":
    print(f"Нәтиже: {a + b}")
elif operation == "-":
    print(f"Нәтиже: {a - b}")
elif operation == "*":
    print(f"Нәтиже: {a * b}")
elif operation == "/":
    if b == 0:
        print("Нөлге бөлуге әрекет жасалды")
    else:
        print(f"Нәтиже: {a / b}")
else:
    print("Белгісіз амал")
````

### Тапсырма 39. Шеңбердің ауданы мен ұзындығы

Шеңбердің радиусын қабылдап, оның ауданы мен ұзындығын қайтаратын `circle_info(radius)` функциясын жазыңыз. Функция екі мәнді бірге қайтаруы керек, ал нәтижелер үтірден кейін екі таңбаға дейін дөңгелектенуі тиіс. π мәні ретінде `3.14159` санын қолданыңыз.

**Күтілетін нәтиже:**

```
circle_info(5) -> (78.54, 31.42)
```

````python
def circle_info(radius):
    pi = 3.14159
    area = pi * (radius ** 2)
    circumference = 2 * pi * radius
    return round(area, 2), round(circumference, 2)
print(circle_info(5))
````

### Тапсырма 41. Палиндромды тексеру

Жол палиндром болса, яғни басынан да, соңынан да бірдей оқылса, `True` қайтаратын `is_palindrome(text)` функциясын жазыңыз. Әріптердің регистрі мен бос орындар ескерілмеуі керек.

**Күтілетін нәтиже:**

```
is_palindrome("Level") -> True
is_palindrome("Was it a car or a cat I saw") -> True
is_palindrome("Python") -> False
```

````python
def is_palindrome(text):
    cleaned_text = text.replace(" ", "").lower()
    return cleaned_text == cleaned_text[::-1]
print(is_palindrome("Level"))                  # True
print(is_palindrome("Was it a car or a cat I saw")) # True
print(is_palindrome("Python"))
````

### Тапсырма 42. Жай сандар

Сан жай сан болса `True` қайтаратын `is_prime(n)` функциясын жазыңыз. Содан кейін осы функцияны қолданып, 1-ден 100-ге дейінгі барлық жай сандардың тізімін және олардың санын шығарыңыз.

**Күтілетін нәтиже:**

```
[2, 3, 5, 7, 11, 13, ..., 97]
Жай сандар саны: 25
```

````python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True
primes = [x for x in range(1, 101) if is_prime(x)]
print(primes)
print(f"Жай сандар саны: {len(primes)}")
````
